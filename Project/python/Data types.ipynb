PROBLEMS

```python
#create variables of type int,float,complex,bool,list,tuple,set,and dict
a = 34   
b = 2.76 
c = 11j  
d = True 
e = [1,2,3,4,5] 
f = (a,b,c,d)
g = {"a1","a2","a3","a4"}
h = {"name":"abc","age":12}
```

```python
# display the data type of each variable using type()
print(type(a))
print(type(b))
print(type(c))
print(type(d))
print(type(e))
print(type(f))
print(type(g))
print(type(h))
```

```python
#create a boolean variable and print its value and type
student = True
print(student)
print(type(student))
```

```python
# check the data type of a user-entered value after converting it to an integer
value = int(input("enter a number"))
print(value)
print(type(value))
```

```python
#create a list, tuple, set, and dictionary, then displays their data types
lst = [23,56,86,12]
tple = ("apple","mango")
st = {"blue","white","red"}
dictn = {"color":"yellow","fruit":"banana"}
print(type(lst))
print(type(tple))
print(type(st))
print(type(dictn))
```

```python
#create variables with different data types and compare their types
ab1 = 78.30
ab2 = "Hello"
print(type(ab1))
print(type(ab2))
print(type(ab1) == type(ab2))
```

```python
#store the same value as an integer, float, and strings, then displays their data types
num_int = 10
num_float = 10.0
num_str = "10"
print(type(num_int))
print(type(num_float))
print(type(num_str))
```

```python
#display the memory address of variables using id
a = 10
b = "Hello"

print(id(a))
print(id(b))
```

```python
#create two variables with the same value and compare their memory addresses
a = 50
b = 50

print(id(a))
print(id(b))
print(id(a) == id(b))
```

```python
#create variables of all basic data types and print their values 
integer = 10
float_num = 10.5
string = "Hello"
boolean = True
complex_num = 2 + 3j

print(integer)
print(float_num)
print(string)
print(boolean)
print(complex_num)
```
