```python
# create and write
with open("word.txt","w") as file:
    file.write("created")
```

```python
# read
with open("word.txt","r") as file:
    print(file.read())
```

```python
# overwrite a file
with open("word.txt","w") as file:
    file.write("welcome to the word")
```

```python
# append new line
with open("word.txt","a") as file:
    file.write("\nHi,This is a word page")
```

```python
with open("word.txt","r") as file:
    print(file.read())
```

```python
# read line by line
with open("word.txt","r") as file:
    for line in file:
        print(line.strip())
```

```python
# read only one line
with open("word.txt","r") as file:
    print(file.readline())
```

```python
# read lines by list
with open("word.txt","r") as file:
    lines = file.readlines()
    print(lines)
```

```python
# working with file pointer
with open("word.txt","r") as file:
    print(file.read(7))
    file.seek(0)
    print(file.read())
```

```python
# current position
with open("word.txt","r") as file:
    print(file.tell())
    file.read(7)
    print(file.tell())
```

```python
#multiple lines
lines = ["It's a page for creating\n", "python is a programming language\n"]
with open("word.txt","w") as file:
    file.writelines(lines)
```

```python
with open("word.txt","r")as file:
    print(file.read())
```

```python
#copy
with open("word.txt","r") as src:
    content = src.read
with open("copy.txt","w") as dest:
    dest.write("content")
    
```

```python
import csv
with open("data.csv","w",newline="") as file:
    value = csv.writer(file)
    value.writerow(["name","age","course"])
    value.writerow(["diya",20,"data science"])
    value.writerow(["john",23,"python"])
    value.writerow(["anu",22,"data analytics"])
```
