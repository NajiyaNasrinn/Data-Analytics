```python
#Create variables of type int, float, str, bool, and complex. Convert each to atleast one other compatible data type and display the original and converted types
num_int = 10
num_float = 10.5
text = "100"
value_bool = True
num_complex = 3 + 4j

int_to_float = float(num_int)
float_to_int = int(num_float)
str_to_int = int(text)
bool_to_int = int(value_bool)
complex_to_str = str(num_complex)

print("Original:", num_int, "Type:", type(num_int))
print("Converted:", int_to_float, "Type:", type(int_to_float))

print("Original:", num_float, "Type:", type(num_float))
print("Converted:", float_to_int, "Type:", type(float_to_int))

print("Original:", text, "Type:", type(text))
print("Converted:", str_to_int, "Type:", type(str_to_int))

print("Original:", value_bool, "Type:", type(value_bool))
print("Converted:", bool_to_int, "Type:", type(bool_to_int))

print("Original:", num_complex, "Type:", type(num_complex))
print("Converted:", complex_to_str, "Type:", type(complex_to_str))
```
