Strings are Arrays

```python
#Print the first character of the string "Python"
char = "Python"
print(char[0])
```

```python
#Print the last character of the string "Programming"
charr = "Programming"
print(charr[-1])
```

```python
#Print the third character of the string"Computer"
charactr = "Computer"
print(charactr[2])
```

```python
#Display the length of the string "Data Science"
string = "Data science"
print(len(string))
```

```python
#Iterate through the string "Welcome" and print each character on a new line
str = "Welcome"
for i in str:
  print(i)
```

Modify Strings

```python
#Convert the string "python programming" to uppercase
s1 = "python programming"
print(s1.upper())
```

```python
#Convert the string "DATA SCIENCE" to lowercase
s2 = "DATA SCIENCE"
print(s2.lower())
```

```python
#Remove the leading and trailing spaces from the string " Hello Python "
a12 = " Hello Python "
print(a12.strip())
```

```python
#Replace "Python" with "Java" in the string "I am learning Python"
ab = "I am learning Python"
a_b = ab.replace("Python" , "Java")
print(a_b)
```

```python
#Split the string "Python is easy to learn" into a list of words.
ab2 = "Python is easy to learn"
ab_2 = ab2.split()
print(ab_2)
```

String Concatenation

```python
#Concatenate the strings "Hello" and "World" with a space
first = "Hello "
second = "world"
print(first + second)
```

```python
#Concatenate the strings "Data" and "Science"
frst = "Data"
scnd = "Science"
print(frst + scnd)
```

```python
#Display a welcome message by concatenating "Welcome" and "Riya"
first1 = "Welcome"
name = "Riya"
print(first1 + " " + name)
```

```python
#Join the strings "Calicut" and "Kerala" using a comma
place = "Calicut"
state = "Kerala"
print(place + "," + state)
```

```python
#Create an email address by concatenating "john" , "doe" , and "@gmail.com"
first_name = "john"
last_name = "doe"
domain = "@gmail.com"
email = first_name + last_name + domain
print(email)
```
