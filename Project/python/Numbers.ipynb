```python
# store an integer and display its value and data type
num_int = 345
print(num_int)
print(type(num_int))
```

```python
#store an floating-point number and display its value and data type
num_float = 789.45
print(num_float)
print(type(num_float))
```

```python
#store an complex number and display its value and data type
num_complex = 56+7j
print(num_complex)
print(type(num_complex))
```

```python
#create variables of type int, float, complex and print them
var = 56
var_1 = 975.2
var_2 = 40+2j
print(var)
print(var_1)
print(var_2)
```

```python
#check whether two variables have the same data type
ab = 234
ab1 = 567
print(type(ab) == type(ab1))
```

```python
#find the sum of an integer and a float
n_int = 50
n_float = 3.50
sum = n_int + n_float
print("sum is:",sum) 
```

```python
#multiply two floating point numbers
float_1 = 3.50
float_2 = 5.2
product = float_1 * float_2
print("Product is:",product)
```

```python
#divide two integers and display the result
int_1 = 450
int_2 = 5
quotient = int_1 / int_2
print("Quotient is:",quotient)
```

```python
#find the square of a number
num = 25
square = num**2  #num*num
print("Square is:",square)
```

```python
#calculate the cube of a number
num1 = 45
cube = num1**3
print("Cube is:",cube)
```
