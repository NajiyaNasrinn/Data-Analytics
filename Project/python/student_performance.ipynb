```python
df["performance"]=df["final_exam_score"].apply(
    lambda x:"Excellent" if x >= 85
    else "Good" if x >= 70
    else "Needs Improvement"
)
```

<h3>INSIGHT :</h3> Students were classified as Excellent, Good, or Needs Improvement. The classification is based on exam scores. It makes performance easier to understand.
