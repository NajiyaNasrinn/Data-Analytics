```python
#20 or below:fail
#21 to 50 pass
#51 to 70 average
# 71 to 80 good
#81 to 89 excellent
# 90 or above outstanding
score = int(input("Enter the score:"))
if score >= 90 and score <= 100:
  print("Outstanding")
elif score >= 80 and score <= 89:
    print("Excellent")
elif score >= 70 and score <= 79:
    print("Good")
elif score >= 50 and score <= 69:
    print("Average")
elif score >= 20 and score <= 49:
    print("pass")
elif score >= 0 and score <= 19:
    print("Fail")    
else:
    print("Invalid")

```
