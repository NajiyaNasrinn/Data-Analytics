```python
# Create an array containing [10, 20, 30, 40]. Create a separate copy and a view of the array. Modify the arrays and observe the effect on the original array.
import numpy as np
a = np.array([10,20,30,40])
b= a.copy()
b[1] = 90
print(a)
print(b)
```

```python
# view 
a1 = np.array([10,20,30,40])
b1 = a1.view()
b1[1] = 90
print(a1)
print(b1)
```
