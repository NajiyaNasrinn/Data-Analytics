Data Integration and Transformation

```python
#Basic Merging
import pandas as pd
df1 = pd.DataFrame({
    "id":[1,2,3],
    "name":["Asha","Binu","Cathy"]
})
df2 = pd.DataFrame({
    "id":[1,2,4],
    "marks":[85,90,75]
})
merged_df = pd.merge(df1,df2,on="id")
print(merged_df)
```

Types of Joins

```python
#inner join
pd.merge(df1,df2,on = "id",how="inner")
```

```python
#left join
pd.merge(df1,df2,on="id",how="left")
```

```python
#right join
pd.merge(df1,df2,on="id",how="right")
```

```python
#outer join
pd.merge(df1,df2,on="id",how="outer")
```

Merge on different column names

```python
import pandas as pd
df1 = pd.DataFrame({
    "id":[1,2,3],
    "name":["Asha","Binu","Cathy"]
})
df2 = pd.DataFrame({
    "student_id":[1,2,4],
    "marks":[85,90,75]
})
merged_df =  pd.merge(df1,df2,left_on="id",right_on="student_id")
merged_df
```

concatenation

```python
#row wise
pd.concat([df1,df2],axis = 0)
```

```python
#column wise
pd.concat([df1,df2],axis=1)
```

Apply

```python
import pandas as pd 
marks = pd.Series([45,60,75,90])
def grade(x):
    if x >= 50:
        return"Pass"
    else:
        return"Fail"
print(marks.apply(grade))
```

Apply using Lambda Function

```python
import pandas as pd
marks = pd.Series([45,60,75,90])
result = marks.apply(lambda x: x + 5)
print(result)
```

```python
# merge with same column name
import pandas as pd
f1 = pd.DataFrame({
    "Id":[1,2,3,4,5],
    "Name":["John","Rahul","Tiya","Rohan","Alia"],
    "Course":["Python","Flutter","Python","Data Science","Flutter"]
})
f2 = pd.DataFrame({
    "Id":[1,2,3,6,7],
    "Marks":[80,94,78,70,90],
    "Duration":["6 mnth","4 mnth","6 mnth","4 mnth","6 mnth"]
})
merged = pd.merge(f1,f2, on="Id")
print(merged)
```

```python
pd.merge(f1,f2,on="Id",how="inner")
```

```python
pd.merge(f1,f2,on="Id",how="left")
```

```python
pd.merge(f1,f2,on="Id",how="right")
```

```python
pd.merge(f1,f2,on="Id",how="outer")
```

```python
# merge with diff column name
import pandas as pd
f11 = pd.DataFrame({
    "Id":[1,2,3,4,5],
    "Name":["John","Rahul","Tiya","Rohan","Alia"],
    "Course":["Python","Flutter","Python","Data Science","Flutter"]
})
f22 = pd.DataFrame({
    "Number":[1,2,3,6,7],
    "Marks":[80,94,78,70,90],
    "Duration":["6 mnth","4 mnth","6 mnth","4 mnth","6 mnth"]
})
merged1 = pd.merge(f11,f22,left_on="Id",right_on="Number")
merged1
```
