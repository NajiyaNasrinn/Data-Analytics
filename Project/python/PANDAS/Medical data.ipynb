```python
df = df.dropna(how="all")
```

```python
df
```

```python
df.columns = df.columns.str.strip()
```

```python
text_columns = df.select_dtypes(include="object").columns
```

```python
for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
df.head()
```

```python
df["Name"] = df["Name"].fillna(df["Name"].mode()[0],inplace=True)
```

```python
df["Gender"]=df["Gender"].fillna(df["Gender"].mode()[0],inplace=True)
```

```python
df["Symptoms"]=df["Symptoms"].fillna(df["Symptoms"].mode()[0],inplace=True)
```

```python
df["Causes"]=df["Causes"].fillna(df["Causes"].mode()[0],inplace=True)
```

```python
df["Disease"]=df["Disease"].fillna(df["Disease"].mode()[0],inplace=True)
```

```python
df["Medicine"]=df["Medicine"].fillna(df["Medicine"].mode()[0],inplace=True)
```

```python
df.isnull().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df["DateOfBirth"] = df["DateOfBirth"].fillna(df["DateOfBirth"].mode()[0],inplace=True)
```

```python
df.isnull().sum().sum()
```

```python
df=df.reset_index(drop=True)
```
